# 📘 01 - Python Lists

Explore how to use lists to store and manipulate data.

## ✅ Learning Goals
- Create and modify Python lists
- Access elements using indexing
- Use list methods like `append()`, `remove()`, and `sort()`

````python
# 🔧 Create a List
my_list = [10, 20, 30, 40, 50]
print(my_list)
````

````python
# 🎯 Access Elements
print('First item:', my_list[0])
print('Last item:', my_list[-1])
````

````python
# ➕ Modify List
my_list.append(60)
my_list.remove(30)
print(my_list)
````

````python
# 🔁 Loop Through List
for item in my_list:
    print(item)
````

## 🔥 Challenge (Optional)
Want to build a list in one line? Python has *list comprehensions* — a compact way to write a loop that builds a list.

```python
squares = [n**2 for n in range(10)]
```

Try it: build a list of the **even** numbers from 1 to 20 using a list comprehension (hint: add an `if` at the end).

````python
squares = []

for n in range(20):
    if n % 2 == 0:
        squares.append[n]
````

````python
for number in squares:
    print(number)
````
